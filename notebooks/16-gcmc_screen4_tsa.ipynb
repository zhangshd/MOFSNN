```python

n_jobs = 64
gases = ["CO2", "N2"]
for temp in [298.0, 323.0, 473.0]:
    dataset = f"stable_coremof_4606_mc{temp}K"
    # dataset = "stable_coremof_4606_mc473.0K.0K"
    gcmc_workdir = Path(f"CGCNN_MT/inference/CoREMOF2019/{dataset}/mc_data")
    result_dir = gcmc_workdir
    ads_dfs = []
    for batch_dir in gcmc_workdir.glob("*"):
        if not batch_dir.is_dir():
            continue
        batch_name = batch_dir.name
        df_ads = process_isotherm_results(batch_dir, gases, unit="mol/kg", verbose=0, n_jobs=n_jobs)
        if len(df_ads) > 0:
            ads_dfs.append(df_ads)
    if len(ads_dfs) == 0:
        continue
    df_ads = pd.concat(ads_dfs, ignore_index=True)
    print("number of isotherm data points: ", len(df_ads))
    if len(df_ads) > 0:
        df_ads.to_csv(result_dir/(f'00-isotherm-data-{dataset}.tsv'), index=False, sep='\t', float_format='%.7f')
# df_ads = process_isotherm_results(gcmc_workdir, gases, unit="mol/kg", verbose=0, n_jobs=n_jobs)
# print("number of isotherm data points: ", len(df_ads))
# if len(df_ads) > 0:
#     df_ads.to_csv(result_dir/(f'00-isotherm-data-{dataset}.tsv'), index=False, sep='\t', float_format='%.6f')
```

```python
root_dir = Path(f"CGCNN_MT/inference/CoREMOF2019")
ads_all = []
valid_cols = ["MofName"]
mix_frac = [0.15, 0.85]
ads_keys = []
for p in [0.1, 1]:
    for gas in ["CO2", "N2"]:
        for temp in [298.0, 323.0, 473.0]:
            dataset = f"stable_coremof_4606_mc{temp}K"
            df_ads = pd.read_csv(root_dir/dataset/'mc_data'/f'00-isotherm-data-{dataset}.tsv', sep='\t')
            ads_col_mix = f"Ads{gas}_{temp}K_{p}bar_mix{mix_frac[0]}"
            ads_col_pure = f"Ads{gas}_{temp}K_{p}bar_pure"

            sub_df_mix = df_ads[(df_ads["Pressure[bar]"]==p)&(df_ads["MoleculeFraction"].isin(mix_frac))&(df_ads["GasName"]==gas)]
            sub_df_pure = df_ads[(df_ads["Pressure[bar]"]==p)&(df_ads["MoleculeFraction"].isin([0, 1]))&(df_ads["GasName"]==gas)]
            sub_df_mix = sub_df_mix.rename(columns={"ExcessLoading": ads_col_mix})
            sub_df_pure = sub_df_pure.rename(columns={"ExcessLoading": ads_col_pure})
            sub_df_merge = pd.merge(sub_df_mix[valid_cols+[ads_col_mix]], 
                                    sub_df_pure[valid_cols+[ads_col_pure]], on=["MofName"], how="outer")
            ads_all.append(sub_df_merge)
            ads_keys.append(ads_col_mix)
            ads_keys.append(ads_col_pure)
ads_all_df = ads_all[0]
for i in range(1, len(ads_all)):
    ads_all_df = pd.merge(ads_all_df, ads_all[i], on=["MofName"], how="outer")
print(ads_all_df.shape)
ads_all_df
```
